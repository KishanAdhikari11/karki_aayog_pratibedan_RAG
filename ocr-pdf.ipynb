```python
import os
import json
import gc
import easyocr
import torch
from pdf2image import convert_from_path
import pypdf


print("Initializing EasyOCR...")
device = True if torch.cuda.is_available() else False
reader = easyocr.Reader(["ne", "en"], gpu=device)


def ocr_pdf_to_json(pdf_path, output_json):
    if not os.path.exists(pdf_path):
        print(f"Error: File not found at {pdf_path}")
        return

    pdf_file = pypdf.PdfReader(pdf_path)
    total_pages = len(pdf_file.pages)
    print(f"Total pages to process: {total_pages}")

    results = []

    for i in range(1, total_pages + 1):
        try:
            print(f"Processing Page {i}/{total_pages}...")

            pages = convert_from_path(pdf_path, first_page=i, last_page=i, dpi=250)

            if not pages:
                continue

            img_array = np.array(pages[0].convert("RGB"))

            text_blocks = reader.readtext(img_array, detail=0, paragraph=True)

            page_content = "\n".join(text_blocks)

            results.append({"page_no": i, "content": page_content})

            del pages
            del img_array
            if device:
                torch.cuda.empty_cache()
            gc.collect()

        except Exception as e:
            print(f"Skipping page {i} due to error: {e}")
            results.append({"page_no": i, "content": "ERROR_PROCESSING_PAGE"})

    with open(output_json, "w", encoding="utf-8") as f:
        json.dump(results, f, ensure_ascii=False, indent=4)

    print(f"\n✅ Success! OCR complete. Data saved to: {output_json}")


INPUT_PDF = "/aayog_report.pdf"
OUTPUT_FILE = "aayog_report_output.json"

ocr_pdf_to_json(INPUT_PDF, OUTPUT_FILE)
```
